```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# Step 1: Read the data
df = pd.read_excel('20240302_OD_Calibration.xlsx')

# Step 2 & 3: Create and fit the model
X = df[['Mean']]  # Feature matrix
y = df['Real_OD']  # Target variable
model = LinearRegression().fit(X, y)

# Step 4: Generate predictions
df['Predicted'] = model.predict(X)

# Step 5: Plot values and predictions
plt.scatter(X, y, color='black', label='Actual values')
plt.plot(X, df['Predicted'], color='blue', linewidth=3, label='Predicted values')
plt.xlabel('Mean')
plt.ylabel('Real_OD')
plt.title('Linear Model: Actual vs. Predicted Real_OD')
plt.legend()
plt.show()
```
