```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# Step 1: Read the data
df = pd.read_excel('../experiments/20240302_OD_Calibration.xlsx')
df.head()
```

```python
# Step 2 & 3: Create and fit the model
X = df[['Mean']]  # Feature matrix
y = df['Real OD']  # Target variable
model = LinearRegression().fit(X, y)

# Step 4: Generate predictions
preds = model.predict(X)
```

```python
# Step 5: Plot values and predictions
plt.scatter(X, y, color='black', label='Actual values')
plt.plot(X.values.flatten(), preds, color='blue', linewidth=3, label='Predicted values')

plt.xlabel('Mean')
plt.ylabel('Real_OD')
plt.title('Linear Model: Actual vs. Predicted Real_OD')
plt.legend()
plt.show()
```

```python
slope = model.coef_[0]  # Assuming there's only one feature
intercept = model.intercept_

slope, intercept
```

```python
import numpy as np
def calculate_growth_rate(inc_vol, run_time, waste_vol): 
    waste_per_hour = waste_vol / run_time
    v_start = inc_vol
    v_end = inc_vol + waste_per_hour 
    gr = v_end / v_start
    return gr

gr = calculate_growth_rate(inc_vol=30, run_time = 10, waste_vol=200)
dt = 1/ np.log2(gr)
print(f'Growth rate per our: {gr:.2f}; Doubling time: {int(dt*60)}min')
```
